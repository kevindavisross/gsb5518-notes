````python
# Model 1: strikes occur independently minute-by-minute, each minute with
# probability 1.5/60 of a strike. There are 180 minutes in 3 hours, so
# X ~ Binomial(n=180, p=1.5/60).

rng = np.random.default_rng(2024)

n_minutes = 180
p_strike = 1.5 / 60
n_sim = 10000

# Simulate X directly as a Binomial random variable
X_sim1 = rng.binomial(n_minutes, p_strike, size=n_sim)

# Simulated distribution: table of values and relative frequencies
values1, counts1 = np.unique(X_sim1, return_counts=True)
sim_probs1 = counts1 / n_sim

# Theoretical distribution: Binomial(180, 1.5/60) pmf
theo_probs1 = stats.binom.pmf(values1, n_minutes, p_strike)

table1 = pd.DataFrame({
    'x': values1,
    'simulated_prob': sim_probs1,
    'theoretical_prob': theo_probs1
})
table1
````

````python
# Plot: simulated relative frequencies vs. theoretical Binomial(180, 1.5/60) pmf
plt.figure()
plt.bar(values1 - 0.15, sim_probs1, width=0.3, label='Simulated')
plt.bar(values1 + 0.15, theo_probs1, width=0.3, label='Theoretical Binomial(180, 1.5/60)')
plt.xlabel('X = number of strikes in 3 hours')
plt.ylabel('Probability')
plt.title('Model 1')
plt.legend()
plt.show()
````

````python
# Approximate mean, variance, and standard deviation from simulation,
# compared to the exact Binomial values
print('Simulated mean, var, sd:   ', X_sim1.mean(), X_sim1.var(), X_sim1.std())

theo_mean1 = n_minutes * p_strike
theo_var1 = n_minutes * p_strike * (1 - p_strike)
print('Theoretical mean, var, sd: ', theo_mean1, theo_var1, np.sqrt(theo_var1))
````

**Hint:** The exponential RVs represent the time between strikes, or *interarrival times*. You can simulate indefinitely many interarrival times (in hours) for many independent realizations of the process at once using

```
interarrival_times = rng.exponential(1 / 1.5, size=(n_sim, max_events))
```

where each row is one realization's sequence of interarrival times $W_0, W_1, W_2, \dots$. Make sure you specify the rate corresponding to time measured in *hours*.

The `count_strikes_in_3_hours` function below takes as input a sequence of interarrival times *measured in hours* (like the $W$'s in the above example) and outputs the number of strikes in the first 3 hours. You can apply this function to each row to get simulated values of $X$, or equivalently, use `np.cumsum` along each row and count how many cumulative interarrival times are at most 3.

````python
def count_strikes_in_3_hours(interarrival_times):
    total_time = 0
    num_strikes = 0
    for time in interarrival_times:
        total_time += time
        if total_time > 3:
            return num_strikes
        else:
            num_strikes += 1
````

````python
# Model 3: interarrival times (hours) are iid Exponential with mean 2/3 hour,
# i.e., rate parameter 1.5 per hour. Simulate many realizations of the sequence
# of interarrival times, and for each realization use count_strikes_in_3_hours
# (or, equivalently, cumulative sums) to determine X.

rate = 1.5
max_events = 200  # comfortably more interarrival times than we could need in 3 hours

interarrival_times = rng.exponential(1 / rate, size=(n_sim, max_events))

# Sanity check: the last cumulative interarrival time in each row should be
# far beyond 3 hours, so we never run out of simulated interarrival times.
cum_times = np.cumsum(interarrival_times, axis=1)
print('Smallest final cumulative time across all simulations:', cum_times[:, -1].min())

# Number of strikes in 3 hours = number of cumulative interarrival times that are at most 3.
# (This gives the same answer as applying count_strikes_in_3_hours to each row.)
X_sim3 = (cum_times <= 3).sum(axis=1)

values3, counts3 = np.unique(X_sim3, return_counts=True)
sim_probs3 = counts3 / n_sim
theo_probs3 = stats.poisson.pmf(values3, 1.5 * 3)

table3 = pd.DataFrame({
    'x': values3,
    'simulated_prob': sim_probs3,
    'theoretical_prob': theo_probs3
})
table3
````

## e) Poisson processes

**In Part e) you should just read and run the cells.  There are a few follow up questions in Part f) that require you to provide your own Python code.**

You should observe that models 1, 2, and 3 all produce similar distributions for $X$.  It turns out that these three different sets of assumptions are equivalent to one another.  For example, if you assume the times between ocurrences are independent and Exponentially distributed as in Model 3, then the counts of events in non-overlapping time intervals will be independent with the count in each interval having a Poisson distribution with appropriate mean.

A random phenomenon which evolves over time according to the assumptions of Model 2 (or equivalently Model 3, or equivalently Model 1) is called a *Poisson process*.  A Poisson process has a single parameter: $\lambda$, the average *rate* at which occurrences happen.  In the example here, $\lambda = 1.5$ strikes per hour on average.  In a Poisson process, the times between occurrences are independent, each having an Exponential distribution with rate parameter $\lambda$.  Also, the number of occurrences that happen in any time interval of length $t$ has a Poisson distribution with mean $\lambda t$.

In the example, we only considered the number of occurrences at one time points, 3 hours.  But in a Poisson process, for every point in time there is a random variable which represents the number of occurrences that have occurred by that time.

We have actually seen an example of a Poisson process before in the textbook; see [Example 2.5](https://kevindavisross.github.io/friendly-probability-book/language-probability.html#exm-PP-outcome), [Example 2.12](https://kevindavisross.github.io/friendly-probability-book/language-probability.html#exm-PP-event) [Example 2.18](https://kevindavisross.github.io/friendly-probability-book/language-probability.html#fig-PP-rv).  There are many random variables we are interested in for a Poisson process: the counts of occurrences at particular times, the times between occurrences, the times at which the occurrences happen.  In Symbulate, defining a `PoissonProcessProbabilitySpace` allows access to all these random variables.  Defining a `RV` `N` on this probability space defines a Poisson process. An outcome can be plotted as a "sample path" which plots the number of occurrences over time.

````python
# A Poisson process is simulated by generating interarrival times and
# accumulating them into arrival times. N(t) counts arrivals by time t.

def simulate_poisson_path(rate, T, rng):
    """Simulate one realization of a Poisson process with the given rate,
    returning the arrival times that fall within [0, T]."""
    arrivals = []
    t = 0
    while True:
        t += rng.exponential(1 / rate)
        if t > T:
            break
        arrivals.append(t)
    return np.array(arrivals)

def plot_poisson_path(arrival_times, T, ax=None, **kwargs):
    ax = ax or plt.gca()
    times = np.concatenate(([0], arrival_times, [T]))
    counts = np.concatenate(([0], np.arange(1, len(arrival_times) + 1), [len(arrival_times)]))
    ax.step(times, counts, where='post', **kwargs)

rng = np.random.default_rng(42)
T_horizon = 10  # hours, just for visualizing sample paths

fig, ax = plt.subplots()
arrivals = simulate_poisson_path(1.5, T_horizon, rng)
plot_poisson_path(arrivals, T_horizon, ax=ax)
ax.set_xlabel('Time (hours)')
ax.set_ylabel('N(t) = number of occurrences by time t')
ax.set_title('One realization of a Poisson process (rate = 1.5)')
plt.show()
````

The horizontal axis represents time (hours, continuous). The vertical axis represents number of occurrences (count, discrete). Refer to the textbook examples mentioned above for a similar example.

Here are a few possible sample paths.

````python
fig, ax = plt.subplots()
for _ in range(5):
    arrivals = simulate_poisson_path(1.5, T_horizon, rng)
    plot_poisson_path(arrivals, T_horizon, ax=ax)
ax.set_xlabel('Time (hours)')
ax.set_ylabel('N(t)')
ax.set_title('5 realizations of the Poisson process (rate = 1.5)')
plt.show()
````

And here are lots

````python
fig, ax = plt.subplots()
for _ in range(100):
    arrivals = simulate_poisson_path(1.5, T_horizon, rng)
    plot_poisson_path(arrivals, T_horizon, ax=ax, alpha=0.3)
ax.set_xlabel('Time (hours)')
ax.set_ylabel('N(t)')
ax.set_title('100 realizations of the Poisson process (rate = 1.5)')
plt.show()
````

For the Poisson process `N`, `N[t]` is the random variable representing the number of occurrences that have happened by time t (hours).  For example `N[3]` is what we called $X$ above.

````python
# For efficiency, simulate many realizations of the process at once by
# generating a matrix of interarrival times (rows = realizations,
# columns = successive interarrival times), then accumulating each row.

def simulate_poisson_process_paths(rate, n_sim, rng, max_events=50):
    """Simulate n_sim independent realizations of a Poisson process.
    Returns (interarrival_times, arrival_times), each an (n_sim, max_events) array,
    where arrival_times is the cumulative sum of interarrival_times along each row."""
    interarrival_times = rng.exponential(1 / rate, size=(n_sim, max_events))
    arrival_times = np.cumsum(interarrival_times, axis=1)
    return interarrival_times, arrival_times

def N_at(t, arrival_times):
    """Number of arrivals by time t, for each realization (row) of arrival_times."""
    return (arrival_times <= t).sum(axis=1)

rng = np.random.default_rng(42)
interarrival_times, arrival_times = simulate_poisson_process_paths(1.5, 10000, rng)

x = N_at(3, arrival_times)
x[:20]  # peek at the first 20 simulated values of N(3)
````

````python
values, counts = np.unique(x, return_counts=True)

plt.figure()
plt.bar(values, counts / len(x), alpha=0.6, label='Simulated N(3)')
k = np.arange(0, values.max() + 1)
plt.plot(k, stats.poisson(1.5 * 3).pmf(k), 'ro-', label='Poisson(1.5*3) pmf')
plt.xlabel('N(3)')
plt.ylabel('Probability')
plt.legend()
plt.show()
````

````python
table = pd.DataFrame({'x': values, 'relative_frequency': counts / len(x)})
table
````
